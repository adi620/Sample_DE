```python
# Building the Gold fact sales table

from pyspark.sql import functions as F

# Read the source tables
sales_df = spark.table("workspace.silver.sales")
customers_df = spark.table("workspace.gold.dim_customers")

# Build fact_sales via a left join, equivalent to the original SQL
fact_sales_df = (
    sales_df.alias("s")
    .join(
        customers_df.alias("c"),
        on=F.col("s.customer_id") == F.col("c.customer_id"),
        how="left",
    )
    .select(
        F.col("s.order_id").alias("order_id"),
        F.col("s.order_date").alias("order_date"),
        F.col("s.product_id").alias("product_id"),
        F.col("s.store_id").alias("store_id"),
        F.col("c.lhid").alias("customer_lhid"),
        F.col("s.quantity").alias("quantity"),
        F.col("s.unit_price").alias("unit_price"),
        F.col("s.discount").alias("discount"),
        F.col("s.revenue").alias("revenue"),
        F.col("s.cost").alias("cost"),
        F.col("s.profit").alias("profit"),
        F.col("s.created_date").alias("created_date"),
    )
)

# Write the result to the gold fact table
fact_sales_df.write.mode("overwrite").saveAsTable("workspace.gold.fact_sales")

display(fact_sales_df)



```
