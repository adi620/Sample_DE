```python
from pyspark.sql import Window
from pyspark.sql.functions import row_number

# Read the customers table from the previous SQL cell
customers_df = spark.read.table("workspace.silver.customers")

# Assign LHID: a monotonically increasing id (1, 2, 3, ...) based on ascending customer_ID
window_spec = Window.orderBy("customer_ID")
customers_df = customers_df.withColumn("LHID", row_number().over(window_spec))

display(customers_df)
```

```python
# Write the customers DataFrame (with LHID) to the gold table
(customers_df.write
    .mode("overwrite")
    .format("delta")
    .saveAsTable("workspace.gold.dim_customers")
)

display(spark.read.table("workspace.gold.dim_customers"))
```
