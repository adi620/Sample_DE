```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.bronze
```

```python
df = spark.read.csv(
    "/Volumes/workspace/landing/sampleproject/customers.csv",
    header=True,
    inferSchema=True
)
from pyspark.sql.functions import current_timestamp

df = df.withColumn("created_date", current_timestamp())

df.write.format("delta").mode("overwrite").saveAsTable("workspace.bronze.customers")
```

```python
df_day2 = spark.read.csv(
    "/Volumes/workspace/landing/sampleproject/customers_day2.csv",
    header=True,
    inferSchema=True
)

df_day2 = df_day2.withColumn("created_date", current_timestamp())

df_day2.write.format("delta").mode("append").saveAsTable("workspace.bronze.customers")
```
